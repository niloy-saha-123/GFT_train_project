# 01 Ingest + QC

Load video, compute metadata/hash, sample frames for quality checks.

```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt

sys.path.append(str(Path.cwd() / 'src'))

from rail_video_intelligence.pipeline.ingest import extract_video_metadata, sample_frames

VIDEO_PATH = Path('data/raw/camera_01_2026_04_09_233235.MOV')
meta = extract_video_metadata(VIDEO_PATH)
meta
```

```python
indices = [0, int(meta.frame_count * 0.25), int(meta.frame_count * 0.5), int(meta.frame_count * 0.75), meta.frame_count - 1]
samples = sample_frames(VIDEO_PATH, indices)

fig, axes = plt.subplots(1, len(samples), figsize=(22, 5))
for ax, (idx, frame) in zip(axes, samples):
    ax.imshow(frame[:, :, ::-1])
    ax.set_title(f'frame {idx}')
    ax.axis('off')
plt.tight_layout()
```
